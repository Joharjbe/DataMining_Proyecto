confirmación de funcionamiento de spark local (más rápido)

```python
from pyspark.sql import SparkSession

spark = (SparkSession.builder
    .appName("Test-Fase0")
    .master("local[2]")
    .config("spark.driver.memory", "2g")
    .config("spark.driver.bindAddress", "127.0.0.1")
    .config("spark.driver.host", "127.0.0.1")
    .getOrCreate())

spark.sparkContext.setLogLevel("WARN")
print(f"Spark versión: {spark.version}")
print(f"Spark UI: {spark.sparkContext.uiWebUrl}")
```

```python
DATASET = "../dataset"

# leemos movies por ser pequeño
movies = spark.read.csv(
    f"{DATASET}/movies.csv",
    header=True,       # primera fila = nombres de columnas
    inferSchema=True   # detecta tipos automáticamente (int, string, etc.)
)

# estructura del DataFrame (columnas y tipos detectados)
movies.printSchema()

# primeras N filas 
movies.show(5, truncate=False)  

print(f"Total películas: {movies.count():,}")
```

cerramos la sesión 

```python
spark.stop()
print("SparkSession local detenida")
```
